```python
import pandas as pd
import numpy as np

# ============================================================
# CASE STUDY 6
# SUPPLY CHAIN & INVENTORY MANAGEMENT
# STEP 1: LOAD SOURCE DATA
# ============================================================

# Load datasets
sales = pd.read_csv('/content/sales_data-2.csv')
inventory = pd.read_csv('/content/inventory_data.csv')
suppliers = pd.read_csv('/content/suppliers_data.csv')
purchase_orders = pd.read_csv('/content/purchase_orders_data.csv')

# ============================================================
# DISPLAY BASIC INFORMATION
# ============================================================

datasets = {
    "Sales": sales,
    "Inventory": inventory,
    "Suppliers": suppliers,
    "Purchase Orders": purchase_orders
}

for name, df in datasets.items():
    print("\n" + "=" * 70)
    print(f"{name.upper()} DATASET")
    print("=" * 70)

    print(f"Rows    : {df.shape[0]}")
    print(f"Columns : {df.shape[1]}")

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nFirst 5 rows:")
    display(df.head())
```

```python
# ============================================================
# CASE STUDY 6
# STEP 2: DATA QUALITY CHECK
# ============================================================

for name, df in datasets.items():

    print("\n" + "=" * 70)
    print(f"{name.upper()} - DATA QUALITY CHECK")
    print("=" * 70)

    # Data types
    print("\n1. DATA TYPES")
    print(df.dtypes)

    # Missing values
    print("\n2. MISSING VALUES")
    missing = df.isnull().sum()
    print(missing)

    # Duplicate rows
    print("\n3. DUPLICATE ROWS")
    print("Duplicate rows:", df.duplicated().sum())

    # Unique values
    print("\n4. UNIQUE VALUES")
    print(df.nunique())
```

```python
# ============================================================
# CASE STUDY 6
# STEP 3: DATE CONVERSION & DATE RANGE VALIDATION
# ============================================================

# Convert date columns to datetime
sales['Sale_Date'] = pd.to_datetime(sales['Sale_Date'], errors='coerce')

inventory['Last_Updated'] = pd.to_datetime(
    inventory['Last_Updated'],
    errors='coerce'
)

purchase_orders['Order_Date'] = pd.to_datetime(
    purchase_orders['Order_Date'],
    errors='coerce'
)

purchase_orders['Arrival_Date'] = pd.to_datetime(
    purchase_orders['Arrival_Date'],
    errors='coerce'
)

# ============================================================
# CHECK FOR INVALID DATES
# ============================================================

print("=" * 70)
print("INVALID DATE CHECK")
print("=" * 70)

print("\nSales invalid dates:",
      sales['Sale_Date'].isna().sum())

print("Inventory invalid dates:",
      inventory['Last_Updated'].isna().sum())

print("Purchase Order invalid order dates:",
      purchase_orders['Order_Date'].isna().sum())

print("Purchase Order invalid arrival dates:",
      purchase_orders['Arrival_Date'].isna().sum())


# ============================================================
# DATE RANGES
# ============================================================

print("\n" + "=" * 70)
print("DATE RANGES")
print("=" * 70)

print("\nSales:")
print("Start Date :", sales['Sale_Date'].min().date())
print("End Date   :", sales['Sale_Date'].max().date())

print("\nInventory:")
print("Start Date :", inventory['Last_Updated'].min().date())
print("End Date   :", inventory['Last_Updated'].max().date())

print("\nPurchase Orders:")
print("Order Start Date :", purchase_orders['Order_Date'].min().date())
print("Order End Date   :", purchase_orders['Order_Date'].max().date())

print("Arrival Start Date :", purchase_orders['Arrival_Date'].min().date())
print("Arrival End Date   :", purchase_orders['Arrival_Date'].max().date())


# ============================================================
# DATE TYPES AFTER CONVERSION
# ============================================================

print("\n" + "=" * 70)
print("DATE DATA TYPES AFTER CONVERSION")
print("=" * 70)

print("\nSales:", sales['Sale_Date'].dtype)
print("Inventory:", inventory['Last_Updated'].dtype)
print("PO Order Date:", purchase_orders['Order_Date'].dtype)
print("PO Arrival Date:", purchase_orders['Arrival_Date'].dtype)
```

```python
# ============================================================
# CASE STUDY 6
# STEP 4: PRIMARY KEY & FOREIGN KEY VALIDATION
# ============================================================

print("=" * 70)
print("PRIMARY KEY VALIDATION")
print("=" * 70)

# ------------------------------------------------------------
# Primary Key / Unique ID checks
# ------------------------------------------------------------

print("\nSALES - Sale_ID")
print("Total rows:", len(sales))
print("Unique Sale_ID:", sales['Sale_ID'].nunique())
print("Duplicate Sale_ID:", sales['Sale_ID'].duplicated().sum())

print("\nPURCHASE ORDERS - Order_ID")
print("Total rows:", len(purchase_orders))
print("Unique Order_ID:", purchase_orders['Order_ID'].nunique())
print("Duplicate Order_ID:", purchase_orders['Order_ID'].duplicated().sum())

print("\nSUPPLIERS - Supplier_ID")
print("Total rows:", len(suppliers))
print("Unique Supplier_ID:", suppliers['Supplier_ID'].nunique())
print("Duplicate Supplier_ID:", suppliers['Supplier_ID'].duplicated().sum())


# ============================================================
# FOREIGN KEY VALIDATION
# ============================================================

print("\n" + "=" * 70)
print("FOREIGN KEY VALIDATION")
print("=" * 70)


# ------------------------------------------------------------
# 1. Product IDs
# ------------------------------------------------------------

sales_products = set(sales['Product_ID'])
inventory_products = set(inventory['Product_ID'])
supplier_products = set(suppliers['Product_ID'])
po_products = set(purchase_orders['Product_ID'])

all_products = sorted(
    sales_products |
    inventory_products |
    supplier_products |
    po_products
)

print("\nPRODUCT_ID VALIDATION")

for product in all_products:
    print(
        product,
        "| Sales:", product in sales_products,
        "| Inventory:", product in inventory_products,
        "| Suppliers:", product in supplier_products,
        "| Purchase Orders:", product in po_products
    )


# ------------------------------------------------------------
# 2. Store IDs
# ------------------------------------------------------------

sales_stores = set(sales['Store_ID'])
inventory_stores = set(inventory['Store_ID'])

print("\nSTORE_ID VALIDATION")

print("Sales Stores:", sorted(sales_stores))
print("Inventory Stores:", sorted(inventory_stores))

print(
    "Stores in Sales but not Inventory:",
    sorted(sales_stores - inventory_stores)
)

print(
    "Stores in Inventory but not Sales:",
    sorted(inventory_stores - sales_stores)
)


# ------------------------------------------------------------
# 3. Supplier IDs
# ------------------------------------------------------------

supplier_ids = set(suppliers['Supplier_ID'])
po_supplier_ids = set(purchase_orders['Supplier_ID'])

print("\nSUPPLIER_ID VALIDATION")

print("Suppliers in Supplier table:", len(supplier_ids))
print("Suppliers in Purchase Orders:", len(po_supplier_ids))

print(
    "PO Supplier IDs missing from Supplier table:",
    sorted(po_supplier_ids - supplier_ids)
)

print(
    "Supplier IDs not used in Purchase Orders:",
    sorted(supplier_ids - po_supplier_ids)
)


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("VALIDATION SUMMARY")
print("=" * 70)

print("\nSales Product IDs not in Inventory:",
      sorted(sales_products - inventory_products))

print("Inventory Product IDs not in Sales:",
      sorted(inventory_products - sales_products))

print("Sales Product IDs not in Suppliers:",
      sorted(sales_products - supplier_products))

print("PO Product IDs not in Suppliers:",
      sorted(po_products - supplier_products))

print("PO Supplier IDs not in Supplier Master:",
      sorted(po_supplier_ids - supplier_ids))
```

```python
# ============================================================
# CASE STUDY 6
# STEP 5: SALES ANALYSIS
# ============================================================

print("=" * 70)
print("SALES ANALYSIS")
print("=" * 70)

# ------------------------------------------------------------
# 1. Overall Sales KPIs
# ------------------------------------------------------------

total_transactions = len(sales)
total_quantity = sales['Quantity_Sold'].sum()
total_revenue = sales['Revenue'].sum()
average_transaction_quantity = sales['Quantity_Sold'].mean()
average_transaction_revenue = sales['Revenue'].mean()

print("\nOVERALL SALES KPIs")
print("-" * 50)

print(f"Total Transactions       : {total_transactions:,}")
print(f"Total Quantity Sold      : {total_quantity:,}")
print(f"Total Revenue            : {total_revenue:,.2f}")
print(f"Avg Quantity / Transaction : {average_transaction_quantity:,.2f}")
print(f"Avg Revenue / Transaction  : {average_transaction_revenue:,.2f}")


# ------------------------------------------------------------
# 2. Product-Level Sales
# ------------------------------------------------------------

product_sales = (
    sales.groupby('Product_ID')
    .agg(
        Transactions=('Sale_ID', 'count'),
        Quantity_Sold=('Quantity_Sold', 'sum'),
        Revenue=('Revenue', 'sum')
    )
    .reset_index()
)

product_sales['Revenue_Per_Unit'] = (
    product_sales['Revenue'] /
    product_sales['Quantity_Sold']
)

product_sales = product_sales.sort_values(
    'Quantity_Sold',
    ascending=False
)

print("\n" + "=" * 70)
print("PRODUCT-LEVEL SALES")
print("=" * 70)

display(product_sales)


# ------------------------------------------------------------
# 3. Store-Level Sales
# ------------------------------------------------------------

store_sales = (
    sales.groupby('Store_ID')
    .agg(
        Transactions=('Sale_ID', 'count'),
        Quantity_Sold=('Quantity_Sold', 'sum'),
        Revenue=('Revenue', 'sum')
    )
    .reset_index()
)

store_sales = store_sales.sort_values(
    'Revenue',
    ascending=False
)

print("\n" + "=" * 70)
print("STORE-LEVEL SALES")
print("=" * 70)

display(store_sales)


# ------------------------------------------------------------
# 4. Monthly Sales
# ------------------------------------------------------------

sales['Year_Month'] = sales['Sale_Date'].dt.to_period('M').astype(str)

monthly_sales = (
    sales.groupby('Year_Month')
    .agg(
        Transactions=('Sale_ID', 'count'),
        Quantity_Sold=('Quantity_Sold', 'sum'),
        Revenue=('Revenue', 'sum')
    )
    .reset_index()
)

monthly_sales = monthly_sales.sort_values('Year_Month')

print("\n" + "=" * 70)
print("MONTHLY SALES")
print("=" * 70)

display(monthly_sales)


# ------------------------------------------------------------
# 5. Fastest & Slowest Products
# ------------------------------------------------------------

fastest_product = product_sales.iloc[0]
slowest_product = product_sales.iloc[-1]

print("\n" + "=" * 70)
print("FASTEST & SLOWEST MOVING PRODUCTS")
print("=" * 70)

print(
    f"\nFastest-moving product by quantity: "
    f"{fastest_product['Product_ID']} "
    f"({fastest_product['Quantity_Sold']:,} units)"
)

print(
    f"Slowest-moving product by quantity: "
    f"{slowest_product['Product_ID']} "
    f"({slowest_product['Quantity_Sold']:,} units)"
)
```

```python
# ============================================================
# CASE STUDY 6
# STEP 6: INVENTORY ANALYSIS
# ============================================================

print("=" * 70)
print("INVENTORY ANALYSIS")
print("=" * 70)

# ------------------------------------------------------------
# 1. Create Inventory Metrics
# ------------------------------------------------------------

inventory['Stock_Gap'] = (
    inventory['Stock_Level'] -
    inventory['Reorder_Level']
)

inventory['Stock_to_Reorder_Ratio'] = (
    inventory['Stock_Level'] /
    inventory['Reorder_Level']
)

# Inventory status
inventory['Stock_Status'] = np.select(
    [
        inventory['Stock_Level'] < inventory['Reorder_Level'],
        inventory['Stock_Level'] == inventory['Reorder_Level'],
        inventory['Stock_Level'] > inventory['Reorder_Level']
    ],
    [
        'Below Reorder Level',
        'At Reorder Level',
        'Above Reorder Level'
    ],
    default='Unknown'
)


# ------------------------------------------------------------
# 2. Overall Inventory KPIs
# ------------------------------------------------------------

total_inventory_records = len(inventory)

below_reorder = (
    inventory['Stock_Status'] == 'Below Reorder Level'
).sum()

at_reorder = (
    inventory['Stock_Status'] == 'At Reorder Level'
).sum()

above_reorder = (
    inventory['Stock_Status'] == 'Above Reorder Level'
).sum()

reorder_risk_pct = (
    below_reorder / total_inventory_records
) * 100

print("\nOVERALL INVENTORY KPIs")
print("-" * 50)

print(f"Total Inventory Records : {total_inventory_records}")
print(f"Below Reorder Level     : {below_reorder}")
print(f"At Reorder Level        : {at_reorder}")
print(f"Above Reorder Level     : {above_reorder}")
print(f"Reorder Risk %          : {reorder_risk_pct:.2f}%")


# ------------------------------------------------------------
# 3. Inventory Status Distribution
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("INVENTORY STATUS DISTRIBUTION")
print("=" * 70)

status_summary = (
    inventory['Stock_Status']
    .value_counts()
    .reset_index()
)

status_summary.columns = [
    'Stock_Status',
    'Records'
]

status_summary['Percentage'] = (
    status_summary['Records'] /
    total_inventory_records
) * 100

display(status_summary)


# ------------------------------------------------------------
# 4. Product-Level Inventory Analysis
# ------------------------------------------------------------

product_inventory = (
    inventory.groupby('Product_ID')
    .agg(
        Inventory_Records=('Product_ID', 'count'),
        Total_Stock=('Stock_Level', 'sum'),
        Average_Stock=('Stock_Level', 'mean'),
        Total_Reorder_Level=('Reorder_Level', 'sum'),
        Average_Reorder_Level=('Reorder_Level', 'mean'),
        Below_Reorder_Count=(
            'Stock_Status',
            lambda x: (x == 'Below Reorder Level').sum()
        )
    )
    .reset_index()
)

product_inventory['Stock_Gap'] = (
    product_inventory['Total_Stock'] -
    product_inventory['Total_Reorder_Level']
)

product_inventory['Reorder_Risk_%'] = (
    product_inventory['Below_Reorder_Count'] /
    product_inventory['Inventory_Records']
) * 100

product_inventory = product_inventory.sort_values(
    'Below_Reorder_Count',
    ascending=False
)

print("\n" + "=" * 70)
print("PRODUCT-LEVEL INVENTORY ANALYSIS")
print("=" * 70)

display(product_inventory)


# ------------------------------------------------------------
# 5. Store-Level Inventory Analysis
# ------------------------------------------------------------

store_inventory = (
    inventory.groupby('Store_ID')
    .agg(
        Inventory_Records=('Product_ID', 'count'),
        Total_Stock=('Stock_Level', 'sum'),
        Average_Stock=('Stock_Level', 'mean'),
        Total_Reorder_Level=('Reorder_Level', 'sum'),
        Below_Reorder_Count=(
            'Stock_Status',
            lambda x: (x == 'Below Reorder Level').sum()
        )
    )
    .reset_index()
)

store_inventory['Reorder_Risk_%'] = (
    store_inventory['Below_Reorder_Count'] /
    store_inventory['Inventory_Records']
) * 100

store_inventory = store_inventory.sort_values(
    'Below_Reorder_Count',
    ascending=False
)

print("\n" + "=" * 70)
print("STORE-LEVEL INVENTORY ANALYSIS")
print("=" * 70)

display(store_inventory)


# ------------------------------------------------------------
# 6. Critical Inventory Records
# ------------------------------------------------------------

critical_inventory = inventory[
    inventory['Stock_Status'] == 'Below Reorder Level'
].sort_values(
    'Stock_Gap'
)

print("\n" + "=" * 70)
print("CRITICAL INVENTORY RECORDS")
print("=" * 70)

display(
    critical_inventory[
        [
            'Product_ID',
            'Store_ID',
            'Warehouse_ID',
            'Stock_Level',
            'Reorder_Level',
            'Stock_Gap',
            'Stock_to_Reorder_Ratio',
            'Last_Updated'
        ]
    ].head(20)
)
```

```python
# ============================================================
# CASE STUDY 6
# STEP 7: SALES + INVENTORY COMBINED ANALYSIS
# ============================================================

print("=" * 70)
print("SALES + INVENTORY COMBINED ANALYSIS")
print("=" * 70)

# ------------------------------------------------------------
# 1. Product-Level Sales Summary
# ------------------------------------------------------------

product_demand = (
    sales.groupby('Product_ID')
    .agg(
        Transactions=('Sale_ID', 'count'),
        Total_Quantity_Sold=('Quantity_Sold', 'sum'),
        Total_Revenue=('Revenue', 'sum'),
        Average_Quantity_Per_Transaction=('Quantity_Sold', 'mean')
    )
    .reset_index()
)


# ------------------------------------------------------------
# 2. Product-Level Inventory Summary
# ------------------------------------------------------------

product_stock = (
    inventory.groupby('Product_ID')
    .agg(
        Inventory_Records=('Product_ID', 'count'),
        Total_Stock=('Stock_Level', 'sum'),
        Average_Stock=('Stock_Level', 'mean'),
        Total_Reorder_Level=('Reorder_Level', 'sum'),
        Below_Reorder_Count=(
            'Stock_Status',
            lambda x: (x == 'Below Reorder Level').sum()
        )
    )
    .reset_index()
)


# ------------------------------------------------------------
# 3. Merge Sales + Inventory
# ------------------------------------------------------------

product_analysis = pd.merge(
    product_demand,
    product_stock,
    on='Product_ID',
    how='inner'
)


# ------------------------------------------------------------
# 4. Derived Metrics
# ------------------------------------------------------------

product_analysis['Stock_Gap'] = (
    product_analysis['Total_Stock'] -
    product_analysis['Total_Reorder_Level']
)

product_analysis['Reorder_Risk_%'] = (
    product_analysis['Below_Reorder_Count'] /
    product_analysis['Inventory_Records']
) * 100

# Average quantity sold per transaction
product_analysis['Avg_Demand_Per_Transaction'] = (
    product_analysis['Total_Quantity_Sold'] /
    product_analysis['Transactions']
)

# Approximate stock-to-demand ratio
product_analysis['Stock_to_Demand_Ratio'] = (
    product_analysis['Total_Stock'] /
    product_analysis['Total_Quantity_Sold']
)

# Inventory coverage proxy
product_analysis['Demand_per_Inventory_Record'] = (
    product_analysis['Total_Quantity_Sold'] /
    product_analysis['Inventory_Records']
)


# ------------------------------------------------------------
# 5. Demand Classification
# ------------------------------------------------------------

median_demand = product_analysis['Total_Quantity_Sold'].median()

product_analysis['Demand_Category'] = np.where(
    product_analysis['Total_Quantity_Sold'] >= median_demand,
    'High Demand',
    'Low Demand'
)


# ------------------------------------------------------------
# 6. Inventory Risk Classification
# ------------------------------------------------------------

product_analysis['Inventory_Risk'] = np.select(
    [
        product_analysis['Reorder_Risk_%'] >= 30,
        product_analysis['Reorder_Risk_%'] >= 20,
        product_analysis['Reorder_Risk_%'] < 20
    ],
    [
        'High Risk',
        'Medium Risk',
        'Low Risk'
    ],
    default='Unknown'
)


# ------------------------------------------------------------
# 7. Combined Business Classification
# ------------------------------------------------------------

product_analysis['Supply_Chain_Category'] = np.select(
    [
        (product_analysis['Demand_Category'] == 'High Demand') &
        (product_analysis['Inventory_Risk'] == 'High Risk'),

        (product_analysis['Demand_Category'] == 'High Demand') &
        (product_analysis['Inventory_Risk'] != 'High Risk'),

        (product_analysis['Demand_Category'] == 'Low Demand') &
        (product_analysis['Inventory_Risk'] == 'High Risk'),

        (product_analysis['Demand_Category'] == 'Low Demand') &
        (product_analysis['Inventory_Risk'] != 'High Risk')
    ],
    [
        'High Demand - High Inventory Risk',
        'High Demand - Lower Inventory Risk',
        'Low Demand - High Inventory Risk',
        'Low Demand - Lower Inventory Risk'
    ],
    default='Unclassified'
)


# ------------------------------------------------------------
# 8. Display Combined Analysis
# ------------------------------------------------------------

product_analysis = product_analysis.sort_values(
    ['Inventory_Risk', 'Total_Quantity_Sold'],
    ascending=[True, False]
)

print("\nPRODUCT-LEVEL SALES + INVENTORY ANALYSIS")
print("=" * 70)

display(product_analysis)


# ------------------------------------------------------------
# 9. Most Critical Products
# ------------------------------------------------------------

critical_products = product_analysis.sort_values(
    ['Below_Reorder_Count', 'Reorder_Risk_%'],
    ascending=[False, False]
)

print("\n" + "=" * 70)
print("PRODUCTS WITH HIGHEST REORDER RISK")
print("=" * 70)

display(
    critical_products[
        [
            'Product_ID',
            'Total_Quantity_Sold',
            'Total_Revenue',
            'Total_Stock',
            'Total_Reorder_Level',
            'Below_Reorder_Count',
            'Reorder_Risk_%',
            'Demand_Category',
            'Inventory_Risk',
            'Supply_Chain_Category'
        ]
    ]
)


# ------------------------------------------------------------
# 10. High Demand + High Risk Products
# ------------------------------------------------------------

high_risk_demand = product_analysis[
    product_analysis['Supply_Chain_Category'] ==
    'High Demand - High Inventory Risk'
]

print("\n" + "=" * 70)
print("HIGH DEMAND + HIGH INVENTORY RISK")
print("=" * 70)

display(high_risk_demand)
```

```python
# ============================================================
# CASE STUDY 6
# STEP 8: PURCHASE ORDER & SUPPLIER LEAD-TIME ANALYSIS
# ============================================================

print("=" * 70)
print("PURCHASE ORDER & SUPPLIER LEAD-TIME ANALYSIS")
print("=" * 70)


# ------------------------------------------------------------
# 1. Calculate Actual Supplier Lead Time
# ------------------------------------------------------------

purchase_orders['Actual_Lead_Time_Days'] = (
    purchase_orders['Arrival_Date'] -
    purchase_orders['Order_Date']
).dt.days


# ------------------------------------------------------------
# 2. Validate Lead-Time Values
# ------------------------------------------------------------

print("\nACTUAL LEAD-TIME SUMMARY")
print("-" * 70)

print(
    purchase_orders['Actual_Lead_Time_Days'].describe()
)


# ------------------------------------------------------------
# 3. Check for Invalid Lead Times
# ------------------------------------------------------------

invalid_lead_time = purchase_orders[
    purchase_orders['Actual_Lead_Time_Days'] < 0
]

print("\nInvalid / Negative Lead-Time Records:",
      len(invalid_lead_time))


# ------------------------------------------------------------
# 4. Purchase Order KPIs
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PURCHASE ORDER KPIs")
print("=" * 70)

print(f"Total Purchase Orders        : {len(purchase_orders):,}")
print(f"Total Quantity Ordered       : {purchase_orders['Quantity'].sum():,}")
print(
    f"Average Order Quantity       : "
    f"{purchase_orders['Quantity'].mean():,.2f}"
)
print(
    f"Average Actual Lead Time    : "
    f"{purchase_orders['Actual_Lead_Time_Days'].mean():.2f} days"
)
print(
    f"Minimum Actual Lead Time    : "
    f"{purchase_orders['Actual_Lead_Time_Days'].min()} days"
)
print(
    f"Maximum Actual Lead Time    : "
    f"{purchase_orders['Actual_Lead_Time_Days'].max()} days"
)


# ------------------------------------------------------------
# 5. Supplier-Level Analysis
# ------------------------------------------------------------

supplier_po_analysis = (
    purchase_orders
    .groupby('Supplier_ID')
    .agg(
        Purchase_Orders=('Order_ID', 'count'),
        Total_Quantity_Ordered=('Quantity', 'sum'),
        Average_Order_Quantity=('Quantity', 'mean'),
        Average_Actual_Lead_Time=('Actual_Lead_Time_Days', 'mean'),
        Minimum_Actual_Lead_Time=('Actual_Lead_Time_Days', 'min'),
        Maximum_Actual_Lead_Time=('Actual_Lead_Time_Days', 'max')
    )
    .reset_index()
)


# ------------------------------------------------------------
# 6. Merge Supplier Master Information
# ------------------------------------------------------------

supplier_master = suppliers[
    [
        'Supplier_ID',
        'Supplier_Name',
        'Product_ID',
        'Lead_Time (days)',
        'Order_Frequency'
    ]
].copy()

supplier_analysis = pd.merge(
    supplier_master,
    supplier_po_analysis,
    on='Supplier_ID',
    how='left'
)


# ------------------------------------------------------------
# 7. Lead-Time Difference
# ------------------------------------------------------------

supplier_analysis['Lead_Time_Difference'] = (
    supplier_analysis['Average_Actual_Lead_Time'] -
    supplier_analysis['Lead_Time (days)']
)


# ------------------------------------------------------------
# 8. Supplier Analysis Output
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SUPPLIER-LEVEL ANALYSIS")
print("=" * 70)

display(
    supplier_analysis.sort_values(
        'Average_Actual_Lead_Time',
        ascending=False
    )
)


# ------------------------------------------------------------
# 9. Supplier Lead-Time Summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("LEAD-TIME SUMMARY BY SUPPLIER")
print("=" * 70)

display(
    supplier_analysis[
        [
            'Supplier_ID',
            'Supplier_Name',
            'Product_ID',
            'Lead_Time (days)',
            'Order_Frequency',
            'Purchase_Orders',
            'Total_Quantity_Ordered',
            'Average_Actual_Lead_Time',
            'Minimum_Actual_Lead_Time',
            'Maximum_Actual_Lead_Time',
            'Lead_Time_Difference'
        ]
    ]
    .sort_values(
        'Average_Actual_Lead_Time',
        ascending=False
    )
)


# ------------------------------------------------------------
# 10. Lead-Time Distribution
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("ACTUAL LEAD-TIME DISTRIBUTION")
print("=" * 70)

lead_time_distribution = (
    purchase_orders['Actual_Lead_Time_Days']
    .value_counts()
    .sort_index()
    .reset_index()
)

lead_time_distribution.columns = [
    'Actual_Lead_Time_Days',
    'Purchase_Order_Count'
]

display(lead_time_distribution)


# ------------------------------------------------------------
# 11. Supplier Order Frequency Distribution
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SUPPLIER ORDER FREQUENCY DISTRIBUTION")
print("=" * 70)

display(
    suppliers['Order_Frequency']
    .value_counts()
    .reset_index()
)


# ------------------------------------------------------------
# 12. Product-Level Procurement Analysis
# ------------------------------------------------------------

product_procurement = (
    purchase_orders
    .groupby('Product_ID')
    .agg(
        Purchase_Orders=('Order_ID', 'count'),
        Total_Quantity_Ordered=('Quantity', 'sum'),
        Average_Order_Quantity=('Quantity', 'mean'),
        Average_Actual_Lead_Time=('Actual_Lead_Time_Days', 'mean')
    )
    .reset_index()
)

print("\n" + "=" * 70)
print("PRODUCT-LEVEL PROCUREMENT ANALYSIS")
print("=" * 70)

display(
    product_procurement.sort_values(
        'Total_Quantity_Ordered',
        ascending=False
    )
)
```

```python
# ============================================================
# CASE STUDY 6
# STEP 10: K-MEANS SUPPLIER CLUSTERING
# ============================================================

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

print("=" * 70)
print("K-MEANS SUPPLIER CLUSTERING")
print("=" * 70)


# ------------------------------------------------------------
# 1. Select Clustering Features
# ------------------------------------------------------------

cluster_features = [
    'Lead_Time (days)',
    'Order_Frequency_Code',
    'Purchase_Orders',
    'Total_Quantity_Ordered',
    'Average_Order_Quantity'
]

X = supplier_cluster[cluster_features].copy()


# ------------------------------------------------------------
# 2. Standardize Features
# ------------------------------------------------------------

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)


# ------------------------------------------------------------
# 3. Test Multiple Values of K
# ------------------------------------------------------------

silhouette_results = []

for k in range(2, 6):

    kmeans_test = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans_test.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels
    )

    silhouette_results.append({
        'Number_of_Clusters': k,
        'Silhouette_Score': score
    })


silhouette_df = pd.DataFrame(
    silhouette_results
)


# ------------------------------------------------------------
# 4. Display Silhouette Scores
# ------------------------------------------------------------

print("\nSILHOUETTE SCORE BY NUMBER OF CLUSTERS")
print("-" * 70)

display(silhouette_df)


# ------------------------------------------------------------
# 5. Select Best K
# ------------------------------------------------------------

best_k = int(
    silhouette_df.loc[
        silhouette_df['Silhouette_Score'].idxmax(),
        'Number_of_Clusters'
    ]
)

best_score = silhouette_df['Silhouette_Score'].max()

print(f"\nSelected Number of Clusters : {best_k}")
print(f"Silhouette Score             : {best_score:.4f}")


# ------------------------------------------------------------
# 6. Final K-Means Model
# ------------------------------------------------------------

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

supplier_cluster['Cluster'] = (
    kmeans.fit_predict(X_scaled) + 1
)


# ------------------------------------------------------------
# 7. Cluster Distribution
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CLUSTER DISTRIBUTION")
print("=" * 70)

cluster_distribution = (
    supplier_cluster['Cluster']
    .value_counts()
    .sort_index()
    .reset_index()
)

cluster_distribution.columns = [
    'Cluster',
    'Supplier_Count'
]

display(cluster_distribution)


# ------------------------------------------------------------
# 8. Supplier Cluster Results
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SUPPLIER CLUSTER ASSIGNMENTS")
print("=" * 70)

display(
    supplier_cluster[
        [
            'Supplier_ID',
            'Supplier_Name',
            'Product_ID',
            'Lead_Time (days)',
            'Order_Frequency',
            'Purchase_Orders',
            'Total_Quantity_Ordered',
            'Average_Order_Quantity',
            'Cluster'
        ]
    ]
    .sort_values(
        ['Cluster', 'Lead_Time (days)']
    )
)


# ------------------------------------------------------------
# 9. Cluster Profile
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CLUSTER PROFILE")
print("=" * 70)

cluster_profile = (
    supplier_cluster
    .groupby('Cluster')
    .agg(
        Supplier_Count=('Supplier_ID', 'count'),
        Average_Lead_Time=('Lead_Time (days)', 'mean'),
        Average_Order_Frequency_Code=(
            'Order_Frequency_Code',
            'mean'
        ),
        Average_Purchase_Orders=(
            'Purchase_Orders',
            'mean'
        ),
        Average_Total_Quantity_Ordered=(
            'Total_Quantity_Ordered',
            'mean'
        ),
        Average_Order_Quantity=(
            'Average_Order_Quantity',
            'mean'
        )
    )
    .reset_index()
)

display(cluster_profile)
```

```python
# ============================================================
# CASE STUDY 6
# STEP 16: FORECASTING MODEL EVALUATION
# ============================================================

import numpy as np

print("=" * 70)
print("DEMAND FORECASTING MODEL EVALUATION")
print("=" * 70)


# ------------------------------------------------------------
# 1. Helper Functions
# ------------------------------------------------------------

def calculate_mae(actual, predicted):
    return np.mean(np.abs(actual - predicted))


def calculate_rmse(actual, predicted):
    return np.sqrt(np.mean((actual - predicted) ** 2))


def calculate_mape(actual, predicted):
    actual = np.asarray(actual)
    predicted = np.asarray(predicted)

    mask = actual != 0

    return np.mean(
        np.abs(
            (actual[mask] - predicted[mask])
            / actual[mask]
        )
    ) * 100


# ------------------------------------------------------------
# 2. Prepare Evaluation Data
# ------------------------------------------------------------

evaluation_results = []

products = sorted(
    forecast_data['Product_ID'].unique()
)

for product in products:

    product_series = (
        forecast_data[
            forecast_data['Product_ID'] == product
        ]
        .sort_values('Month')
        .set_index('Month')['Quantity_Sold']
    )

    train_series = product_series[
        product_series.index <= pd.Timestamp('2036-08-01')
    ]

    test_series = product_series[
        product_series.index >= pd.Timestamp('2036-09-01')
    ]


    # --------------------------------------------------------
    # 3. Generate Forecasts for Test Period
    # --------------------------------------------------------

    actual = test_series.values


    # 3-Month Moving Average
    ma3_forecast = []

    for i in range(len(test_series)):

        historical_values = product_series[
            product_series.index < test_series.index[i]
        ].tail(3)

        ma3_forecast.append(
            historical_values.mean()
        )


    # 6-Month Moving Average
    ma6_forecast = []

    for i in range(len(test_series)):

        historical_values = product_series[
            product_series.index < test_series.index[i]
        ].tail(6)

        ma6_forecast.append(
            historical_values.mean()
        )


    # 12-Month Seasonal Naive
    seasonal_forecast = []

    for test_month in test_series.index:

        previous_year_month = (
            test_month - pd.DateOffset(years=1)
        )

        seasonal_value = product_series[
            product_series.index == previous_year_month
        ]

        if len(seasonal_value) > 0:
            seasonal_forecast.append(
                seasonal_value.iloc[0]
            )
        else:
            seasonal_forecast.append(np.nan)


    # --------------------------------------------------------
    # 4. Evaluate Each Method
    # --------------------------------------------------------

    model_forecasts = {
        '3-Month Moving Average': np.array(ma3_forecast),
        '6-Month Moving Average': np.array(ma6_forecast),
        '12-Month Seasonal Naive': np.array(seasonal_forecast)
    }


    for model_name, predictions in model_forecasts.items():

        valid_mask = ~np.isnan(predictions)

        actual_valid = actual[valid_mask]
        predicted_valid = predictions[valid_mask]

        evaluation_results.append({
            'Product_ID': product,
            'Model': model_name,
            'MAE': calculate_mae(
                actual_valid,
                predicted_valid
            ),
            'RMSE': calculate_rmse(
                actual_valid,
                predicted_valid
            ),
            'MAPE_%': calculate_mape(
                actual_valid,
                predicted_valid
            )
        })


# ------------------------------------------------------------
# 5. Create Evaluation DataFrame
# ------------------------------------------------------------

evaluation_results_df = pd.DataFrame(
    evaluation_results
)


# ------------------------------------------------------------
# 6. Display Product-Level Results
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PRODUCT-LEVEL MODEL PERFORMANCE")
print("=" * 70)

display(
    evaluation_results_df.round(2)
)


# ------------------------------------------------------------
# 7. Average Performance Across Products
# ------------------------------------------------------------

model_comparison = (
    evaluation_results_df
    .groupby('Model')
    .agg(
        Average_MAE=('MAE', 'mean'),
        Average_RMSE=('RMSE', 'mean'),
        Average_MAPE=('MAPE_%', 'mean')
    )
    .reset_index()
    .sort_values('Average_MAE')
)


print("\n" + "=" * 70)
print("OVERALL MODEL COMPARISON")
print("=" * 70)

display(
    model_comparison.round(2)
)


# ------------------------------------------------------------
# 8. Best-Performing Model by Product
# ------------------------------------------------------------

best_model_by_product = (
    evaluation_results_df
    .sort_values(
        ['Product_ID', 'MAE']
    )
    .groupby('Product_ID')
    .first()
    .reset_index()
)


print("\n" + "=" * 70)
print("LOWEST-MAE MODEL BY PRODUCT")
print("=" * 70)

display(
    best_model_by_product[
        [
            'Product_ID',
            'Model',
            'MAE',
            'RMSE',
            'MAPE_%'
        ]
    ].round(2)
)
```

```python
# ============================================================
# CASE STUDY 6
# STEP 18: REORDER POINT CALCULATION
# ============================================================

print("=" * 70)
print("REORDER POINT CALCULATION")
print("=" * 70)


# ------------------------------------------------------------
# 1. Product-Level Historical Demand Statistics
# ------------------------------------------------------------

demand_statistics = (
    forecast_data
    .groupby('Product_ID')
    .agg(
        Average_Monthly_Demand=('Quantity_Sold', 'mean'),
        Demand_Std_Dev_Monthly=('Quantity_Sold', 'std'),
        Total_Historical_Demand=('Quantity_Sold', 'sum')
    )
    .reset_index()
)


# ------------------------------------------------------------
# 2. Convert Monthly Demand to Approximate Daily Demand
# ------------------------------------------------------------

demand_statistics['Average_Daily_Demand'] = (
    demand_statistics['Average_Monthly_Demand'] / 30
)


# ------------------------------------------------------------
# 3. Calculate Product-Level Supplier Lead Time
# ------------------------------------------------------------

product_lead_time = (
    suppliers
    .groupby('Product_ID')
    .agg(
        Supplier_Count=('Supplier_ID', 'nunique'),
        Average_Lead_Time_Days=('Lead_Time (days)', 'mean'),
        Minimum_Lead_Time_Days=('Lead_Time (days)', 'min'),
        Maximum_Lead_Time_Days=('Lead_Time (days)', 'max')
    )
    .reset_index()
)


# ------------------------------------------------------------
# 4. Merge Demand and Lead-Time Information
# ------------------------------------------------------------

reorder_point_data = demand_statistics.merge(
    product_lead_time,
    on='Product_ID',
    how='left'
)


# ------------------------------------------------------------
# 5. Calculate Demand During Lead Time
# ------------------------------------------------------------

reorder_point_data['Lead_Time_Demand'] = (
    reorder_point_data['Average_Daily_Demand']
    *
    reorder_point_data['Average_Lead_Time_Days']
)


# ------------------------------------------------------------
# 6. Estimate Daily Demand Standard Deviation
# ------------------------------------------------------------

reorder_point_data['Demand_Std_Dev_Daily'] = (
    reorder_point_data['Demand_Std_Dev_Monthly']
    / np.sqrt(30)
)


# ------------------------------------------------------------
# 7. Calculate Safety Stock
# ------------------------------------------------------------
# Service-level assumption:
# 95% service level -> Z = 1.645

Z_VALUE = 1.645

reorder_point_data['Safety_Stock'] = (
    Z_VALUE
    *
    reorder_point_data['Demand_Std_Dev_Daily']
    *
    np.sqrt(
        reorder_point_data['Average_Lead_Time_Days']
    )
)


# ------------------------------------------------------------
# 8. Calculate Reorder Point
# ------------------------------------------------------------

reorder_point_data['Reorder_Point'] = (
    reorder_point_data['Lead_Time_Demand']
    +
    reorder_point_data['Safety_Stock']
)


# ------------------------------------------------------------
# 9. Round Operational Values
# ------------------------------------------------------------

numeric_columns = [
    'Average_Monthly_Demand',
    'Demand_Std_Dev_Monthly',
    'Average_Daily_Demand',
    'Average_Lead_Time_Days',
    'Minimum_Lead_Time_Days',
    'Maximum_Lead_Time_Days',
    'Lead_Time_Demand',
    'Demand_Std_Dev_Daily',
    'Safety_Stock',
    'Reorder_Point'
]

reorder_point_data[numeric_columns] = (
    reorder_point_data[numeric_columns]
    .round(2)
)


# ------------------------------------------------------------
# 10. Display Supplier Lead-Time Information
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PRODUCT-LEVEL SUPPLIER LEAD TIME")
print("=" * 70)

display(
    product_lead_time.round(2)
)


# ------------------------------------------------------------
# 11. Display Reorder Point Calculation
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("REORDER POINT RESULTS")
print("=" * 70)

display(
    reorder_point_data[
        [
            'Product_ID',
            'Average_Monthly_Demand',
            'Average_Daily_Demand',
            'Average_Lead_Time_Days',
            'Lead_Time_Demand',
            'Safety_Stock',
            'Reorder_Point'
        ]
    ]
)


# ------------------------------------------------------------
# 12. Display Formula Components
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("REORDER POINT FORMULA")
print("=" * 70)

print("Reorder Point = Lead-Time Demand + Safety Stock")
print("Lead-Time Demand = Average Daily Demand × Average Lead Time")
print("Safety Stock = Z × Daily Demand Std. Dev. × √Lead Time")
print("Service Level Assumption = 95%")
print("Z-Value = 1.645")
```

```python
# ============================================================
# CASE STUDY 6
# STEP 19: INVENTORY VS CALCULATED REORDER POINT
# ============================================================

print("=" * 70)
print("INVENTORY VS CALCULATED REORDER POINT")
print("=" * 70)


# ------------------------------------------------------------
# 1. Prepare Product-Level Reorder Point
# ------------------------------------------------------------

product_reorder_points = (
    reorder_point_data[
        [
            'Product_ID',
            'Average_Monthly_Demand',
            'Average_Lead_Time_Days',
            'Lead_Time_Demand',
            'Safety_Stock',
            'Reorder_Point'
        ]
    ]
    .copy()
)


# ------------------------------------------------------------
# 2. Aggregate Inventory by Product and Store
# ------------------------------------------------------------

inventory_product_store = (
    inventory
    .groupby(
        ['Product_ID', 'Store_ID'],
        as_index=False
    )
    .agg(
        Inventory_Records=('Product_ID', 'count'),
        Current_Stock=('Stock_Level', 'sum'),
        Existing_Reorder_Level=('Reorder_Level', 'sum')
    )
)


# ------------------------------------------------------------
# 3. Merge Calculated Reorder Point
# ------------------------------------------------------------

inventory_reorder_analysis = inventory_product_store.merge(
    product_reorder_points,
    on='Product_ID',
    how='left'
)


# ------------------------------------------------------------
# 4. Calculate Store-Level Reorder Requirement
# ------------------------------------------------------------

inventory_reorder_analysis['Calculated_Reorder_Point'] = (
    inventory_reorder_analysis['Reorder_Point']
)


inventory_reorder_analysis['Stock_vs_Calculated_Reorder'] = (
    inventory_reorder_analysis['Current_Stock']
    -
    inventory_reorder_analysis['Calculated_Reorder_Point']
)


inventory_reorder_analysis['Additional_Stock_Required'] = (
    inventory_reorder_analysis['Calculated_Reorder_Point']
    -
    inventory_reorder_analysis['Current_Stock']
).clip(lower=0)


inventory_reorder_analysis['Replenishment_Status'] = np.where(
    inventory_reorder_analysis['Current_Stock']
    <
    inventory_reorder_analysis['Calculated_Reorder_Point'],
    'Below Calculated Reorder Point',
    'At/Above Calculated Reorder Point'
)


# ------------------------------------------------------------
# 5. Round Values
# ------------------------------------------------------------

round_columns = [
    'Current_Stock',
    'Existing_Reorder_Level',
    'Average_Monthly_Demand',
    'Average_Lead_Time_Days',
    'Lead_Time_Demand',
    'Safety_Stock',
    'Calculated_Reorder_Point',
    'Stock_vs_Calculated_Reorder',
    'Additional_Stock_Required'
]

inventory_reorder_analysis[round_columns] = (
    inventory_reorder_analysis[round_columns]
    .round(2)
)


# ------------------------------------------------------------
# 6. Display Complete Analysis
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PRODUCT-STORE REORDER ANALYSIS")
print("=" * 70)

display(
    inventory_reorder_analysis[
        [
            'Product_ID',
            'Store_ID',
            'Current_Stock',
            'Existing_Reorder_Level',
            'Calculated_Reorder_Point',
            'Stock_vs_Calculated_Reorder',
            'Additional_Stock_Required',
            'Replenishment_Status'
        ]
    ]
)


# ------------------------------------------------------------
# 7. Replenishment Summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("REPLENISHMENT SUMMARY")
print("=" * 70)

replenishment_summary = (
    inventory_reorder_analysis
    .groupby('Replenishment_Status')
    .agg(
        Product_Store_Records=('Product_ID', 'count'),
        Total_Current_Stock=('Current_Stock', 'sum'),
        Total_Additional_Stock_Required=(
            'Additional_Stock_Required',
            'sum'
        )
    )
    .reset_index()
)

display(
    replenishment_summary.round(2)
)


# ------------------------------------------------------------
# 8. Count Replenishment Records
# ------------------------------------------------------------

below_calculated = (
    inventory_reorder_analysis[
        inventory_reorder_analysis['Replenishment_Status']
        ==
        'Below Calculated Reorder Point'
    ]
)

print(
    f"\nRecords Below Calculated Reorder Point : "
    f"{len(below_calculated)}"
)

print(
    f"Records At/Above Calculated Reorder Point : "
    f"{len(inventory_reorder_analysis) - len(below_calculated)}"
)


# ------------------------------------------------------------
# 9. Highest Replenishment Requirements
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("HIGHEST ADDITIONAL STOCK REQUIREMENTS")
print("=" * 70)

display(
    inventory_reorder_analysis
    .sort_values(
        'Additional_Stock_Required',
        ascending=False
    )
    .head(15)[
        [
            'Product_ID',
            'Store_ID',
            'Current_Stock',
            'Calculated_Reorder_Point',
            'Additional_Stock_Required',
            'Replenishment_Status'
        ]
    ]
)
```

```python
# ============================================================
# CASE STUDY 6
# STEP 20: EXISTING VS CALCULATED REORDER LEVEL
# ============================================================

print("=" * 70)
print("EXISTING VS CALCULATED REORDER LEVEL")
print("=" * 70)


# ------------------------------------------------------------
# 1. Aggregate Existing Inventory Reorder Levels by Product
# ------------------------------------------------------------

existing_reorder_by_product = (
    inventory
    .groupby('Product_ID')
    .agg(
        Inventory_Records=('Product_ID', 'count'),
        Total_Current_Stock=('Stock_Level', 'sum'),
        Average_Current_Stock=('Stock_Level', 'mean'),
        Existing_Reorder_Level=('Reorder_Level', 'mean'),
        Total_Reorder_Level=('Reorder_Level', 'sum')
    )
    .reset_index()
)


# ------------------------------------------------------------
# 2. Prepare Calculated Reorder Point
# ------------------------------------------------------------

calculated_rop = (
    reorder_point_data[
        [
            'Product_ID',
            'Average_Monthly_Demand',
            'Average_Lead_Time_Days',
            'Lead_Time_Demand',
            'Safety_Stock',
            'Reorder_Point'
        ]
    ]
    .copy()
    .rename(
        columns={
            'Reorder_Point': 'Calculated_Reorder_Point'
        }
    )
)


# ------------------------------------------------------------
# 3. Merge Existing and Calculated Reorder Levels
# ------------------------------------------------------------

reorder_level_comparison = existing_reorder_by_product.merge(
    calculated_rop,
    on='Product_ID',
    how='left'
)


# ------------------------------------------------------------
# 4. Calculate Difference
# ------------------------------------------------------------

reorder_level_comparison['Reorder_Level_Difference'] = (
    reorder_level_comparison['Calculated_Reorder_Point']
    -
    reorder_level_comparison['Existing_Reorder_Level']
)


reorder_level_comparison['Difference_Percentage'] = (
    reorder_level_comparison['Reorder_Level_Difference']
    /
    reorder_level_comparison['Existing_Reorder_Level']
) * 100


# ------------------------------------------------------------
# 5. Classify the Difference
# ------------------------------------------------------------

reorder_level_comparison['Threshold_Comparison'] = np.select(
    [
        reorder_level_comparison['Reorder_Level_Difference'] > 0,
        reorder_level_comparison['Reorder_Level_Difference'] < 0
    ],
    [
        'Calculated ROP Higher',
        'Calculated ROP Lower'
    ],
    default='Same'
)


# ------------------------------------------------------------
# 6. Round Numeric Values
# ------------------------------------------------------------

numeric_columns = [
    'Total_Current_Stock',
    'Average_Current_Stock',
    'Existing_Reorder_Level',
    'Total_Reorder_Level',
    'Average_Monthly_Demand',
    'Average_Lead_Time_Days',
    'Lead_Time_Demand',
    'Safety_Stock',
    'Calculated_Reorder_Point',
    'Reorder_Level_Difference',
    'Difference_Percentage'
]

reorder_level_comparison[numeric_columns] = (
    reorder_level_comparison[numeric_columns]
    .round(2)
)


# ------------------------------------------------------------
# 7. Product-Level Comparison
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PRODUCT-LEVEL REORDER LEVEL COMPARISON")
print("=" * 70)

display(
    reorder_level_comparison[
        [
            'Product_ID',
            'Existing_Reorder_Level',
            'Calculated_Reorder_Point',
            'Reorder_Level_Difference',
            'Difference_Percentage',
            'Threshold_Comparison'
        ]
    ]
)


# ------------------------------------------------------------
# 8. Summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("THRESHOLD COMPARISON SUMMARY")
print("=" * 70)

comparison_summary = (
    reorder_level_comparison
    .groupby('Threshold_Comparison')
    .agg(
        Product_Count=('Product_ID', 'count'),
        Average_Difference=('Reorder_Level_Difference', 'mean'),
        Average_Difference_Percentage=('Difference_Percentage', 'mean')
    )
    .reset_index()
)

display(
    comparison_summary.round(2)
)


# ------------------------------------------------------------
# 9. Current Stock vs Both Thresholds
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CURRENT STOCK VS BOTH REORDER THRESHOLDS")
print("=" * 70)

display(
    reorder_level_comparison[
        [
            'Product_ID',
            'Average_Current_Stock',
            'Existing_Reorder_Level',
            'Calculated_Reorder_Point',
            'Reorder_Level_Difference'
        ]
    ]
)
```

```python
# ============================================================
# CASE STUDY 6
# STEP 21: PRODUCT-LEVEL INVENTORY OPTIMIZATION ANALYSIS
# ============================================================

print("=" * 70)
print("PRODUCT-LEVEL INVENTORY OPTIMIZATION ANALYSIS")
print("=" * 70)


# ------------------------------------------------------------
# 1. Create Next-Month Forecast DataFrame
# ------------------------------------------------------------

forecast_next_month = pd.DataFrame({
    'Product_ID': ['P001', 'P002', 'P003', 'P004', 'P005'],
    'Forecast_Quantity': [1170, 1373, 1345, 1852, 1466],
    'Forecast_Method': [
        '3-Month Moving Average',
        '6-Month Moving Average',
        '6-Month Moving Average',
        '3-Month Moving Average',
        '3-Month Moving Average'
    ]
})


# ------------------------------------------------------------
# 2. Aggregate Current Inventory by Product
# ------------------------------------------------------------

product_inventory = (
    inventory
    .groupby('Product_ID')
    .agg(
        Inventory_Records=('Product_ID', 'count'),
        Total_Current_Stock=('Stock_Level', 'sum'),
        Average_Current_Stock=('Stock_Level', 'mean'),
        Existing_Reorder_Level=('Reorder_Level', 'mean')
    )
    .reset_index()
)


# ------------------------------------------------------------
# 3. Prepare Calculated Reorder Point
# ------------------------------------------------------------

calculated_reorder = (
    reorder_point_data[
        [
            'Product_ID',
            'Average_Daily_Demand',
            'Average_Lead_Time_Days',
            'Lead_Time_Demand',
            'Safety_Stock',
            'Reorder_Point'
        ]
    ]
    .copy()
)


# ------------------------------------------------------------
# 4. Merge Inventory + Forecast + Reorder Point
# ------------------------------------------------------------

inventory_optimization = (
    product_inventory
    .merge(
        forecast_next_month,
        on='Product_ID',
        how='left'
    )
    .merge(
        calculated_reorder,
        on='Product_ID',
        how='left'
    )
)


# ------------------------------------------------------------
# 5. Calculate Inventory Metrics
# ------------------------------------------------------------

inventory_optimization['Stock_After_Forecast'] = (
    inventory_optimization['Total_Current_Stock']
    -
    inventory_optimization['Forecast_Quantity']
)


inventory_optimization['Stock_vs_ROP'] = (
    inventory_optimization['Total_Current_Stock']
    -
    inventory_optimization['Reorder_Point']
)


inventory_optimization['Forecast_vs_ROP'] = (
    inventory_optimization['Forecast_Quantity']
    -
    inventory_optimization['Reorder_Point']
)


# ------------------------------------------------------------
# 6. Inventory Status
# ------------------------------------------------------------

inventory_optimization['Inventory_Status'] = np.select(
    [
        inventory_optimization['Total_Current_Stock']
        <
        inventory_optimization['Reorder_Point'],

        inventory_optimization['Stock_After_Forecast']
        <
        0
    ],
    [
        'Below Calculated ROP',
        'Forecast Exceeds Current Stock'
    ],
    default='Above Calculated ROP'
)


# ------------------------------------------------------------
# 7. Round Numeric Values
# ------------------------------------------------------------

numeric_columns = [
    'Total_Current_Stock',
    'Average_Current_Stock',
    'Existing_Reorder_Level',
    'Forecast_Quantity',
    'Average_Daily_Demand',
    'Average_Lead_Time_Days',
    'Lead_Time_Demand',
    'Safety_Stock',
    'Reorder_Point',
    'Stock_After_Forecast',
    'Stock_vs_ROP',
    'Forecast_vs_ROP'
]

inventory_optimization[numeric_columns] = (
    inventory_optimization[numeric_columns]
    .round(2)
)


# ------------------------------------------------------------
# 8. Display Product-Level Analysis
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PRODUCT-LEVEL INVENTORY OPTIMIZATION")
print("=" * 70)

display(
    inventory_optimization[
        [
            'Product_ID',
            'Total_Current_Stock',
            'Existing_Reorder_Level',
            'Reorder_Point',
            'Forecast_Quantity',
            'Stock_After_Forecast',
            'Stock_vs_ROP',
            'Inventory_Status'
        ]
    ]
)


# ------------------------------------------------------------
# 9. Summary Statistics
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("INVENTORY OPTIMIZATION SUMMARY")
print("=" * 70)

print(
    f"Total Products Analyzed : "
    f"{len(inventory_optimization)}"
)

print(
    f"Total Current Stock : "
    f"{inventory_optimization['Total_Current_Stock'].sum():,.0f}"
)

print(
    f"Total Next-Month Forecast : "
    f"{inventory_optimization['Forecast_Quantity'].sum():,.0f}"
)

print(
    f"Total Stock After Forecast : "
    f"{inventory_optimization['Stock_After_Forecast'].sum():,.0f}"
)


# ------------------------------------------------------------
# 10. Inventory Status Distribution
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("INVENTORY STATUS DISTRIBUTION")
print("=" * 70)

status_summary = (
    inventory_optimization
    .groupby('Inventory_Status')
    .size()
    .reset_index(name='Product_Count')
)

display(status_summary)


# ------------------------------------------------------------
# 11. Forecast vs Current Stock
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FORECAST VS CURRENT STOCK")
print("=" * 70)

display(
    inventory_optimization[
        [
            'Product_ID',
            'Total_Current_Stock',
            'Forecast_Quantity',
            'Stock_After_Forecast'
        ]
    ]
    .sort_values(
        'Stock_After_Forecast'
    )
)
```

```python
# ============================================================
# CASE STUDY 6
# STEP 22: FINAL PRODUCT-LEVEL SUPPLY CHAIN ANALYTICS DATASET
# ============================================================

print("=" * 70)
print("FINAL PRODUCT-LEVEL SUPPLY CHAIN ANALYTICS DATASET")
print("=" * 70)


# ------------------------------------------------------------
# 1. Product-Level Sales Summary
# ------------------------------------------------------------

product_sales_summary = (
    sales
    .groupby('Product_ID')
    .agg(
        Total_Transactions=('Sale_ID', 'count'),
        Total_Quantity_Sold=('Quantity_Sold', 'sum'),
        Total_Revenue=('Revenue', 'sum'),
        Average_Quantity_Per_Transaction=('Quantity_Sold', 'mean')
    )
    .reset_index()
)


# ------------------------------------------------------------
# 2. Product-Level Inventory Summary
# ------------------------------------------------------------

product_inventory_summary = (
    inventory
    .groupby('Product_ID')
    .agg(
        Inventory_Records=('Product_ID', 'count'),
        Total_Current_Stock=('Stock_Level', 'sum'),
        Average_Current_Stock=('Stock_Level', 'mean'),
        Existing_Reorder_Level=('Reorder_Level', 'mean'),
        Below_Reorder_Records=(
            'Stock_Level',
            lambda x: 0
        )
    )
    .reset_index()
)

# Calculate actual below-reorder records separately
below_reorder_counts = (
    inventory
    .assign(
        Below_Reorder=lambda x:
        x['Stock_Level'] < x['Reorder_Level']
    )
    .groupby('Product_ID')['Below_Reorder']
    .sum()
    .reset_index(name='Below_Reorder_Records')
)

product_inventory_summary = (
    product_inventory_summary
    .drop(columns='Below_Reorder_Records')
    .merge(
        below_reorder_counts,
        on='Product_ID',
        how='left'
    )
)


# ------------------------------------------------------------
# 3. Supplier Lead-Time Summary
# ------------------------------------------------------------

product_supplier_summary = (
    suppliers
    .groupby('Product_ID')
    .agg(
        Supplier_Count=('Supplier_ID', 'nunique'),
        Average_Supplier_Lead_Time=(
            'Lead_Time (days)',
            'mean'
        ),
        Minimum_Supplier_Lead_Time=(
            'Lead_Time (days)',
            'min'
        ),
        Maximum_Supplier_Lead_Time=(
            'Lead_Time (days)',
            'max'
        )
    )
    .reset_index()
)


# ------------------------------------------------------------
# 4. Purchase Order Summary
# ------------------------------------------------------------

product_po_summary = (
    purchase_orders
    .groupby('Product_ID')
    .agg(
        Total_Purchase_Orders=('Order_ID', 'count'),
        Total_Quantity_Ordered=('Quantity', 'sum'),
        Average_Order_Quantity=('Quantity', 'mean')
    )
    .reset_index()
)


# ------------------------------------------------------------
# 5. Prepare Forecast Data
# ------------------------------------------------------------

final_forecast = forecast_next_month[
    [
        'Product_ID',
        'Forecast_Quantity',
        'Forecast_Method'
    ]
].copy()


# ------------------------------------------------------------
# 6. Prepare Reorder Point Data
# ------------------------------------------------------------

final_rop = reorder_point_data[
    [
        'Product_ID',
        'Average_Monthly_Demand',
        'Average_Daily_Demand',
        'Demand_Std_Dev_Monthly',
        'Average_Lead_Time_Days',
        'Lead_Time_Demand',
        'Safety_Stock',
        'Reorder_Point'
    ]
].copy()


# ------------------------------------------------------------
# 7. Merge All Product-Level Components
# ------------------------------------------------------------

final_product_analytics = (
    product_sales_summary
    .merge(
        product_inventory_summary,
        on='Product_ID',
        how='left'
    )
    .merge(
        product_supplier_summary,
        on='Product_ID',
        how='left'
    )
    .merge(
        product_po_summary,
        on='Product_ID',
        how='left'
    )
    .merge(
        final_forecast,
        on='Product_ID',
        how='left'
    )
    .merge(
        final_rop,
        on='Product_ID',
        how='left'
    )
)


# ------------------------------------------------------------
# 8. Calculate Additional KPIs
# ------------------------------------------------------------

final_product_analytics['Revenue_Per_Unit'] = (
    final_product_analytics['Total_Revenue']
    /
    final_product_analytics['Total_Quantity_Sold']
)


final_product_analytics['Stock_After_Forecast'] = (
    final_product_analytics['Total_Current_Stock']
    -
    final_product_analytics['Forecast_Quantity']
)


final_product_analytics['Stock_vs_ROP'] = (
    final_product_analytics['Total_Current_Stock']
    -
    final_product_analytics['Reorder_Point']
)


final_product_analytics['Forecast_vs_ROP'] = (
    final_product_analytics['Forecast_Quantity']
    -
    final_product_analytics['Reorder_Point']
)


final_product_analytics['Inventory_Risk_Percentage'] = (
    final_product_analytics['Below_Reorder_Records']
    /
    final_product_analytics['Inventory_Records']
) * 100


# ------------------------------------------------------------
# 9. Demand Classification
# ------------------------------------------------------------

median_demand = (
    final_product_analytics['Total_Quantity_Sold']
    .median()
)

final_product_analytics['Demand_Category'] = np.where(
    final_product_analytics['Total_Quantity_Sold']
    >= median_demand,
    'High Demand',
    'Lower Demand'
)


# ------------------------------------------------------------
# 10. Reorder Risk Classification
# ------------------------------------------------------------

median_inventory_risk = (
    final_product_analytics['Inventory_Risk_Percentage']
    .median()
)

final_product_analytics['Reorder_Risk_Category'] = np.where(
    final_product_analytics['Inventory_Risk_Percentage']
    >= median_inventory_risk,
    'Higher Risk',
    'Lower Risk'
)


# ------------------------------------------------------------
# 11. Final Inventory Status
# ------------------------------------------------------------

final_product_analytics['Final_Inventory_Status'] = np.where(
    final_product_analytics['Total_Current_Stock']
    <
    final_product_analytics['Reorder_Point'],
    'Below Calculated ROP',
    'Above Calculated ROP'
)


# ------------------------------------------------------------
# 12. Round Numeric Columns
# ------------------------------------------------------------

numeric_columns = [
    'Total_Quantity_Sold',
    'Total_Revenue',
    'Average_Quantity_Per_Transaction',
    'Total_Current_Stock',
    'Average_Current_Stock',
    'Existing_Reorder_Level',
    'Average_Supplier_Lead_Time',
    'Minimum_Supplier_Lead_Time',
    'Maximum_Supplier_Lead_Time',
    'Total_Quantity_Ordered',
    'Average_Order_Quantity',
    'Forecast_Quantity',
    'Average_Monthly_Demand',
    'Average_Daily_Demand',
    'Demand_Std_Dev_Monthly',
    'Average_Lead_Time_Days',
    'Lead_Time_Demand',
    'Safety_Stock',
    'Reorder_Point',
    'Revenue_Per_Unit',
    'Stock_After_Forecast',
    'Stock_vs_ROP',
    'Forecast_vs_ROP',
    'Inventory_Risk_Percentage'
]

final_product_analytics[numeric_columns] = (
    final_product_analytics[numeric_columns]
    .round(2)
)


# ------------------------------------------------------------
# 13. Display Final Dataset
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL PRODUCT-LEVEL ANALYTICS")
print("=" * 70)

display(
    final_product_analytics
    .sort_values(
        'Total_Quantity_Sold',
        ascending=False
    )
)


# ------------------------------------------------------------
# 14. Dataset Validation
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL DATASET VALIDATION")
print("=" * 70)

print(
    f"Number of Products : "
    f"{final_product_analytics['Product_ID'].nunique()}"
)

print(
    f"Number of Rows : "
    f"{len(final_product_analytics)}"
)

print(
    f"Missing Values : "
    f"{final_product_analytics.isna().sum().sum()}"
)

print(
    f"Duplicate Rows : "
    f"{final_product_analytics.duplicated().sum()}"
)
```
